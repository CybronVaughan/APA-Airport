```python
import pandas as pd
```

```python
dataframe = pd.read_csv('../data/metaheuristics_dataset.csv', sep=';', index_col=False, dtype=str).fillna('').astype(str)
```

```python
def round_3(value):
    return round(value, 3)
```

```python
def type_float(value):
    if(value == ''):
        return ''
    else:
        return float(value)
```

```python
def type_float_round_3(value):
    if(value == ''):
        return ''
    else:
        return round_3(float(value))
```

```python
dataframe['Iterations'] = dataframe['Iterations'].astype(int)
dataframe['Average cost'] = dataframe['Average cost'].astype(float).apply(round_3)
dataframe['Best cost'] = dataframe['Best cost'].astype(int)
dataframe['Average performance time(s)'] = dataframe['Average performance time(s)'].apply(type_float)
dataframe['Best performance time(s)'] = dataframe['Best performance time(s)'].astype(float)
dataframe['GAP(%)'] = dataframe['GAP(%)'].astype(float).apply(round_3)
dataframe['Average GAP(%)'] = dataframe['Average GAP(%)'].apply(type_float_round_3)
```
